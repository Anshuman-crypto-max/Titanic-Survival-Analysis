```python
df['age']=df['age'].fillna(df['age'].median())
```

```python
df
```

```python
df['embarked']=df['embarked'].fillna(df['embarked'].mode()[0])
```

```python
df
```

```python
df.drop_duplicates(inplace=True)
```

```python
df
```

```python
df[['age', 'fare']].boxplot()
```

```python
Q1 = df['fare'].quantile(0.25)
Q3 = df['fare'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df['fare'] >= lower) & (df['fare'] <= upper)]
```
